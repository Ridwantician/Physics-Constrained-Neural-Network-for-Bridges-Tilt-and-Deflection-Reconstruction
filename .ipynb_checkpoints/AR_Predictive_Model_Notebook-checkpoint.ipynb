# AR(1) Cross-Sensor Predictive Model
## IDA-KI OpenLab Research Bridge, TU Dresden — Middle T-girder

### Overview
This notebook implements a next-day tiltmeter prediction model using an autoregressive AR(1) architecture with cross-sensor coupling. It predicts tiltx at both sensor positions (x = 11 m and x = 19 m) one day ahead, using Ridge regression with 10 input features.

### Model Equation
```
tiltx(t) = a1·tiltx_11(t-1) + a2·tiltx_19(t-1) + b·T(t) + c·ΔT(t) 
           + d1·dT_11(t) + d2·dT_19(t) + e_phase + f·T(t)·phase4 + intercept − bias
```

### Bridge Context
- **Structure:** 2 × 15 m continuous prestressed concrete beam, middle T-girder
- **Sensors:** Two tiltmeters at x = 11 m (span AB) and x = 19 m (span BC)
- **Observations:** 253 daily averages from Feb – Oct 2024, four loading phases
- **Train/Test:** 213 days (Feb 22 – Sep 21) / 40 days (Sep 22 – Oct 31)

### Key Design Decisions
1. **Cross-sensor coupling:** Both tiltx_11(t-1) and tiltx_19(t-1) are used because the spans are structurally coupled through the interior support (r = −0.83 between sensors)
2. **Ridge regression (α = 0.1):** Regularises the 10 correlated features and prevents overfitting on 213 samples
3. **14-day bias correction:** Compensates for slow baseline drift (creep, prestress relaxation) that fixed phase coefficients cannot track
4. **phase4 × T interaction:** Allows Phase 4 to have its own thermal sensitivity — critical because the test period is entirely Phase 4

### Why AR Outperforms LSTM/GRU
The tiltx signal is a near-unit-root linear process (lag-1 autocorrelation r = 0.994) with only 213 training samples. The AR model (10 parameters, r = 0.84) outperforms LSTM (hidden=32, ~13k parameters, r = 0.61) and GRU (hidden=16, ~3.5k parameters, r = 0.58) because the signal-to-noise ratio is too low for deep models to learn meaningful nonlinear structure from so few samples.

### Output
This model produces a 40-day tiltx prediction used as input to the AR-PCNN pipeline for full-field deflection prediction.

Standard scientific Python stack plus scikit-learn for Ridge regression. Key constants:
- `REF_DATE`: Feb 1, 2024 — all day indices are relative to this date
- `CP` / `LP`: Phase colours and labels used consistently across all thesis plots
- `OUT`: Output directory for this model's results

````python
================================================================================
AR PREDICTIVE MODEL: AR(1) CROSS-SENSOR TILTX FORECASTING
================================================================================
IDA-KI OpenLab Research Bridge, TU Dresden — Middle T-girder

Predicts next-day tiltx at x=11m and x=19m from:
  - Yesterday's tiltx at BOTH sensors (cross-sensor coupling)
  - Today's temperature, temperature change, front-back gradients
  - Loading phase (one-hot encoded)

MODEL:
  tiltx(t) = a1*tiltx_11(t-1) + a2*tiltx_19(t-1) + b*T(t)
             + c*T_diff(t) + d1*dT_11(t) + d2*dT_19(t) + e_phase
             + f*T(t)*phase4

  Fitted with Ridge regression (alpha=0.1) for regularisation.
  14-day bias correction applied to test predictions.

TRAIN/TEST:
  Train: Feb 22 – Sep 21 (213 days, measured)
  Test:  Sep 22 – Oct 31 (40 days, predicted)

OUTPUTS:
  - ar_predictions.csv: actual vs predicted tiltx for test period
  - ar_summary.json: model coefficients, metrics
  - 4 plots: predictions, error distribution, coefficients, residual analysis
================================================================================
"""

import numpy as np, pandas as pd
import matplotlib; matplotlib.use('Agg')
import matplotlib.pyplot as plt
from sklearn.linear_model import Ridge
import os, json
from datetime import datetime, timedelta

OUT = 'outputs/processing/Tiltx/Standalone_AR'
os.makedirs(OUT, exist_ok=True)
REF_DATE = pd.Timestamp('2024-02-01').date()

CP = {1: '#2196F3', 2: '#4CAF50', 3: '#FF9800', 4: '#F44336'}
LP = {1: 'Ph1', 2: 'Ph2: +tracks', 3: 'Ph3: +veh4.1t', 4: 'Ph4: +veh9t'}
````

## 2. Utility: Day-to-Date Conversion

Converts a day index (relative to Feb 1, 2024) to a human-readable date string for plot labels and JSON output.

````python
def day_to_date(d):
    return (pd.Timestamp('2024-02-01') + timedelta(days=int(d))).strftime('%B %d')
````

## 3. Data Loading and Preprocessing

This function performs the following steps:
1. **Read raw CSV:** Sub-daily tiltmeter and temperature measurements
2. **Compute mean temperature:** T = (T_front + T_back) / 2 at x = 11 m
3. **Daily averaging:** Aggregate all sub-daily readings to daily means for tiltx, temperature, and thermal gradients
4. **Filter:** Keep only days ≥ 21 (first 20 days are commissioning/unstable)
5. **Phase assignment:** Based on known installation dates:
   - Phase 1 (day 21–45): Dead load only
   - Phase 2 (day 46–80): Railway tracks installed  
   - Phase 3 (day 81–147): Test vehicle 4.1 t
   - Phase 4 (day 148+): Test vehicle increased to 9 t
6. **Derived features:** Front-back gradients (dT_11, dT_19), day-to-day temperature change (T_diff)
7. **Date column:** For time series plotting

````python
def load_data():
    raw = pd.read_csv('cleaned_tilt_data.csv')
    raw['Timestamp'] = pd.to_datetime(raw['Timestamp'])
    raw['T'] = (raw['temp_at_11m_front'] + raw['temp_at_11m_back']) / 2
    raw['day'] = (raw['Timestamp'].dt.date - REF_DATE).apply(lambda x: x.days)
    daily = raw.groupby('day').agg({
        'tiltx_11m': 'mean', 'tiltx_19m': 'mean', 'T': 'mean',
        'change_in_temp_11m': 'mean', 'change_in_temp_19m': 'mean',
    }).reset_index()
    daily = daily[daily['day'] >= 21].copy().reset_index(drop=True)
    daily['phase'] = 1
    daily.loc[daily['day'] >= 46, 'phase'] = 2
    daily.loc[daily['day'] >= 81, 'phase'] = 3
    daily.loc[daily['day'] >= 148, 'phase'] = 4
    daily['dT_11'] = daily['change_in_temp_11m']
    daily['dT_19'] = daily['change_in_temp_19m']
    daily['T_diff'] = daily['T'].diff().fillna(0)
    daily['date'] = daily['day'].apply(
        lambda d: pd.Timestamp('2024-02-01') + timedelta(days=int(d)))
    for ph in [1, 2, 3, 4]:
        print(f"  Phase {ph}: {(daily['phase'] == ph).sum()} days")
    return daily
````

## 4. AR Feature Matrix Construction

Builds the 10-feature input matrix for the AR(1) model. For each prediction target at day `t`, the features are:

| Index | Feature | Physical Meaning |
|-------|---------|-----------------|
| 0 | tiltx_11m(t−1) | Yesterday's tilt at 11 m (self-coupling) |
| 1 | tiltx_19m(t−1) | Yesterday's tilt at 19 m (cross-sensor coupling) |
| 2 | T(t) | Today's mean surface temperature |
| 3 | T_diff(t) | Temperature change from yesterday: T(t) − T(t−1) |
| 4 | dT_11(t) | Front-back thermal gradient at 11 m |
| 5 | dT_19(t) | Front-back thermal gradient at 19 m |
| 6 | phase_2 | One-hot: tracks installed |
| 7 | phase_3 | One-hot: vehicle 4.1 t |
| 8 | phase_4 | One-hot: vehicle 9 t |
| 9 | phase4 × T | Phase-specific thermal sensitivity |

**Why phase4 × T?** Removing this interaction collapses the 19 m R² from 0.521 to −0.689. Phase 4 has a fundamentally different thermal sensitivity because the heavier vehicle changes the beam's curvature-temperature relationship.

````python
def build_ar_features(df, start, end):
    """Build 9-feature matrix for AR(1) cross-sensor model.
    Row i uses data from day (start+i-1) to predict day (start+i).

    Features:
      0: tiltx_11m(t-1)   — autoregressive, self
      1: tiltx_19m(t-1)   — autoregressive, cross-sensor
      2: T(t)             — today's temperature (from forecast in deployment)
      3: T_diff(t)        — temperature change from yesterday
      4: dT_11(t)         — front-back gradient at 11m
      5: dT_19(t)         — front-back gradient at 19m
      6: phase_2          — one-hot
      7: phase_3          — one-hot
      8: phase_4          — one-hot
      9: phase_4 × T(t)   — phase-specific thermal sensitivity
    """
    T = df['T'].values[start:end]
    ph = df['phase'].values[start:end]
    return np.column_stack([
        df['tiltx_11m'].values[start - 1:end - 1],
        df['tiltx_19m'].values[start - 1:end - 1],
        T,
        df['T_diff'].values[start:end],
        df['dT_11'].values[start:end],
        df['dT_19'].values[start:end],
        (ph == 2).astype(float),
        (ph == 3).astype(float),
        (ph == 4).astype(float),
        T * (ph == 4).astype(float),
    ])
````

## 5. Accuracy Metrics Function

Computes six metrics between actual and predicted values:
- **R²** (coefficient of determination): 1 = perfect, 0 = mean predictor, <0 = worse than mean
- **RMSE** (root mean square error): in physical units (mm/m)
- **MAE** (mean absolute error): more robust to outliers than RMSE
- **MAPE** (mean absolute percentage error): relative accuracy (excluded when values near zero)
- **Bias** (mean signed error): positive = systematic overestimation
- **r** (Pearson correlation): captures shape agreement regardless of offset/scale

````python
def met(y, yp):
    err = yp - y
    ss = np.sum(err**2); st = np.sum((y - y.mean())**2)
    mask = np.abs(y) > 1e-6
    mape = float(np.mean(np.abs(err[mask] / y[mask])) * 100) if mask.sum() > 0 else float('nan')
    r = float(np.corrcoef(y.ravel(), yp.ravel())[0, 1]) if y.ravel().std() > 1e-10 else 0.0
    return {
        'R2': float(1 - ss / (st + 1e-12)),
        'RMSE': float(np.sqrt(np.mean(err**2))),
        'MAE': float(np.mean(np.abs(err))),
        'MAPE': mape,
        'Bias': float(np.mean(err)),
        'r': r,
        'Rel_L2': float(np.sqrt(ss) / (np.sqrt(np.sum(y**2)) + 1e-12)),
    }
````

## 7. Train AR Models with Bias Correction

Two separate Ridge regression models are fitted — one for each sensor location. The procedure:

1. **Fit:** Ridge(α=0.1) on the 10-feature training matrix
2. **Predict raw:** Apply the model to test features
3. **Bias correction:** 
   - Compute mean prediction residual on the last 14 training days
   - Subtract this bias estimate from all test predictions
   - This compensates for slow baseline drift that fixed phase coefficients cannot track
4. **Evaluate:** Compute metrics for both raw and bias-corrected predictions

### Why 14-Day Window?
The bias correction window must be:
- **Long enough** to average out daily noise (14 days ≈ 2 weeks of thermal cycles)
- **Short enough** to capture recent drift trends
- **Within training period** so it doesn't use test data

### Physical Interpretation of Coefficients
- **Self-coupling (a1 ≈ 0.81):** Yesterday's tilt at the same sensor explains ~66% of today's variance (0.81² ≈ 0.66). This high persistence reflects the thermal inertia of massive concrete (slow heating/cooling)
- **Cross-coupling (a2 ≈ 0.15):** The other sensor contributes via structural coupling through the interior support. Sign reflects the anti-correlation (r = −0.83) between spans
- **All non-AR coefficients flip sign** between the 11 m and 19 m models — physically expected because the two sensors are on opposite sides of the interior support
- **T(t) and T_diff(t) ≈ 0:** The AR term already encodes thermal state via yesterday's tiltx, making explicit temperature redundant
- **phase4 × T ≈ 0 but critical:** Near-zero coefficient but removing it collapses the 19 m R² from 0.521 to −0.689

````python
TRAIN AR MODELS
# ═══════════════════════════════════════════════════════════════
print(f"\n{'='*70}")
print("TRAINING AR(1) CROSS-SENSOR MODELS")
print(f"{'='*70}")

X_train = build_ar_features(daily, 1, N_TRAIN)
X_test = build_ar_features(daily, N_TRAIN, N)

feat_names = ['tiltx_11m(t-1)', 'tiltx_19m(t-1)', 'T(t)', 'T_diff(t)',
              'dT_11(t)', 'dT_19(t)', 'phase_2', 'phase_3', 'phase_4', 'phase_4×T']

results = {}
BIAS_WINDOW = 14  # last 14 training days for bias estimation
for col in ['tiltx_11m', 'tiltx_19m']:
    y_train = daily[col].values[1:N_TRAIN]
    y_test = daily[col].values[N_TRAIN:]

    model = Ridge(alpha=0.1).fit(X_train, y_train)
    pred_train = model.predict(X_train)
    pred_test_raw = model.predict(X_test)

    # Bias correction: estimate systematic offset from last BIAS_WINDOW training days
    X_last = build_ar_features(daily, N_TRAIN - BIAS_WINDOW, N_TRAIN)
    y_last = daily[col].values[N_TRAIN - BIAS_WINDOW:N_TRAIN]
    bias_est = float(np.mean(model.predict(X_last) - y_last))
    pred_test = pred_test_raw - bias_est

    m_train = met(y_train, pred_train)
    m_test_raw = met(y_test, pred_test_raw)
    m_test = met(y_test, pred_test)

    label = '11m' if '11' in col else '19m'
    results[label] = {
        'model': model,
        'pred_train': pred_train, 'actual_train': y_train,
        'pred_test': pred_test, 'pred_test_raw': pred_test_raw,
        'actual_test': y_test,
        'met_train': m_train, 'met_test': m_test, 'met_test_raw': m_test_raw,
        'bias_correction': bias_est,
        'coefs': dict(zip(feat_names, model.coef_)),
        'intercept': float(model.intercept_),
    }

    print(f"\n  {label}:")
    print(f"    Train: RMSE={m_train['RMSE']:.4f} R2={m_train['R2']:.4f} r={m_train['r']:.4f}")
    print(f"    Test (raw):      RMSE={m_test_raw['RMSE']:.4f} R2={m_test_raw['R2']:.4f} "
          f"Bias={m_test_raw['Bias']:+.4f}")
    print(f"    Test (corrected): RMSE={m_test['RMSE']:.4f} R2={m_test['R2']:.4f} "
          f"Bias={m_test['Bias']:+.4f} r={m_test['r']:.4f}")
    print(f"    Bias correction ({BIAS_WINDOW}-day window): {bias_est:+.6f} mm/m")
    print(f"    Coefficients:")
    for fn, c in zip(feat_names, model.coef_):
        print(f"      {fn:>20s}: {c:+.6f}")
    print(f"      {'intercept':>20s}: {model.intercept_:+.6f}")

# Physical interpretation
ar_self_11 = results['11m']['coefs']['tiltx_11m(t-1)']
ar_cross_11 = results['11m']['coefs']['tiltx_19m(t-1)']
print(f"\n  Physical interpretation:")
print(f"    AR self-coeff (11m): {ar_self_11:.4f} → yesterday explains "
      f"{ar_self_11**2*100:.0f}% of today's variance")
print(f"    AR cross-coeff (11m→from 19m): {ar_cross_11:.4f}")
print(f"    High AR coeff reflects slow thermal response of massive concrete")

# ═══════════════════════════════════════════════════════════════
#
````

## 13. Plot 5: Test Set Metrics Bar Chart

Bar chart showing R², RMSE, MAE, and Bias at both sensor positions for the bias-corrected test period. The 11 m sensor has better predictability (R² = 0.730) than the 19 m sensor (R² = 0.521) because span AB (fixed-fixed) has a more constrained and predictable thermal response than span BC (fixed-pinned).

````python
# ── Plot 5: Metrics bar chart (R2, RMSE, MAE, MAPE) ──
fig, axes = plt.subplots(1, 5, figsize=(25, 5))
fig.suptitle('AR(1) Test Set Metrics', fontsize=13, fontweight='bold')
metrics_to_plot = [('R2', 'R²'), ('RMSE', 'RMSE (mm/m)'), ('MAE', 'MAE (mm/m)'), ('Bias', 'Bias (mm/m)'), ('Rel_L2', 'Relative L² Error')]
for ax, (mkey, mlabel) in zip(axes, metrics_to_plot):
    vals = [results[l]['met_test'][mkey] for l in ['11m', '19m']]
    bars = ax.bar(['11m', '19m'], vals, color=['steelblue', 'coral'], alpha=0.8, edgecolor='k')
    for bar, v in zip(bars, vals):
        ax.annotate(f'{v:.4f}' if mkey != 'Bias' else f'{v:+.4f}',
                    xy=(bar.get_x() + bar.get_width()/2, bar.get_height()),
                    ha='center', va='bottom', fontsize=10, fontweight='bold')
    ax.set_ylabel(mlabel); ax.set_title(mlabel)
    ax.grid(True, alpha=0.3, axis='y')
    if mkey == 'R2':
        ax.set_ylim(0, 1.1)
        ax.axhline(1.0, color='green', ls='--', lw=1, alpha=0.5, label='Perfect')
        ax.legend(fontsize=7)
plt.tight_layout()
plt.savefig(os.path.join(OUT, '05_metrics_bar.png'), dpi=200); plt.close()
print("  Saved: 05_metrics_bar.png")
````
